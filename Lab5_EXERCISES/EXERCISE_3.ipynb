```python
import random
import time

LAST_NAME = "Mabanglo"
STUDENT_ID = "TUPM-26-1847"
FAVORITE_ARTIST = "Zara Larsson"

SEED_DIGIT = int(STUDENT_ID[-1])

ID_SUM = sum(
    int(character)
    for character in STUDENT_ID
    if character.isdigit()
)

NAME_LENGTH = len(LAST_NAME)

ARTIST_SUM = sum(
    ord(character.upper()) - ord("A") + 1
    for character in FAVORITE_ARTIST
    if character.isalpha()
)

# Generate One Common Order Dataset

order_count = 12

orders = [
    ID_SUM * 1000
    + ARTIST_SUM
    + NAME_LENGTH
    + (index * 23)
    for index in range(order_count)
]

sorted_orders = sorted(orders)
reverse_orders = list(reversed(sorted_orders))

random_orders = sorted_orders.copy()

random_generator = random.Random(
    ID_SUM + ARTIST_SUM + SEED_DIGIT
)

random_generator.shuffle(random_orders)

print("=== GENERATED ORDER DATASET ===")
print("ORIGINAL ORDERS:", orders)
print("SORTED INPUT:", sorted_orders)
print("REVERSE INPUT:", reverse_orders)
print("RANDOM INPUT:", random_orders)
```

```python
rotated_records = [7, 8, 9, 1, 2, 3, 4, 5, 6]


# Bubble Sort
def bubble_sort(values):
    data = values.copy()
    comparisons = 0
    movements = 0

    for outer_index in range(len(data)):
        swapped = False

        for inner_index in range(
            0,
            len(data) - outer_index - 1
        ):
            comparisons += 1

            if data[inner_index] > data[inner_index + 1]:
                data[inner_index], data[inner_index + 1] = (
                    data[inner_index + 1],
                    data[inner_index]
                )

                movements += 1
                swapped = True

        if not swapped:
            break

    return data, comparisons, movements


# Run Bubble Sort
sorted_records, sort_comparisons, sort_movements = bubble_sort(
    rotated_records
)

print("=== BUBBLE SORT ===")
print("ORIGINAL RECORDS:", rotated_records)
print("SORTED RECORDS:", sorted_records)
print("COMPARISONS:", sort_comparisons)
print("MOVEMENTS:", sort_movements)
```

```python
def selection_sort(values):
    data = values.copy()
    comparisons = 0
    movements = 0

    for outer_index in range(len(data)):
        min_index = outer_index

        for inner_index in range(outer_index + 1, len(data)):
            comparisons += 1

            if data[inner_index] < data[min_index]:
                min_index = inner_index

        if min_index != outer_index:
            data[outer_index], data[min_index] = (
                data[min_index],
                data[outer_index]
            )
            movements += 1

    return data, comparisons, movements


# Sample Records
records = [64, 25, 12, 22, 11]

# Run Selection Sort
sorted_records, sort_comparisons, sort_movements = selection_sort(records)

# Display Results
print("=== SELECTION SORT ===")
print("ORIGINAL RECORDS:", records)
print("SORTED RECORDS:", sorted_records)
print("COMPARISONS:", sort_comparisons)
print("MOVEMENTS:", sort_movements)
```

```python
import time

def insertion_sort(values):
    data = values.copy()
    comparisons = 0
    movements = 0

    for index in range(1, len(data)):
        key = data[index]
        position = index

        while position > 0:
            comparisons += 1

            if data[position - 1] > key:
                data[position] = data[position - 1]
                movements += 1
                position -= 1
            else:
                break

        if position != index:
            data[position] = key
            movements += 1

    return data, comparisons, movements


sorting_algorithms = {
    "BUBBLE SORT": bubble_sort,
    "SELECTION SORT": selection_sort,
    "INSERTION SORT": insertion_sort
}

input_conditions = {
    "SORTED": sorted_orders,
    "REVERSE": reverse_orders,
    "RANDOM": random_orders
}

results = {}

for condition_name, input_data in input_conditions.items():
    results[condition_name] = {}

    for algorithm_name, algorithm in sorting_algorithms.items():
        start_time = time.perf_counter()

        output, comparisons, movements = algorithm(input_data)

        end_time = time.perf_counter()

        results[condition_name][algorithm_name] = {
            "output": output,
            "comparisons": comparisons,
            "movements": movements,
            "execution_time": end_time - start_time,
            "sorted_correctly": output == sorted_orders
        }

print("=== SORTING PERFORMANCE RESULTS ===")

for condition_name, algorithms in results.items():
    print("\nINPUT CONDITION:", condition_name)

    for algorithm_name, result in algorithms.items():
        print("\n", algorithm_name)
        print("OUTPUT:", result["output"])
        print("COMPARISONS:", result["comparisons"])
        print("MOVEMENTS:", result["movements"])
        print(
            "EXECUTION TIME:",
            f"{result['execution_time']:.10f}",
            "seconds"
        )
        print("SORTING VERIFIED:", result["sorted_correctly"])
```

```python
print("=== FINAL E-COMMERCE ORDER PROCESSING REPORT ===")
print("GENERATED ORDER DATASET:", orders)
print("SORTED INPUT:", sorted_orders)
print("REVERSE INPUT:", reverse_orders)
print("RANDOM INPUT:", random_orders)

for condition_name in results:
    print("\n", condition_name, "INPUT RESULTS")

for algorithm_name in results[condition_name]:
        result = results[condition_name][algorithm_name]

print(
            algorithm_name,
            "| Comparisons:",
            result["comparisons"],
            "| Movements:",
            result["movements"],
            "| Time:",
            f"{result['execution_time']:.10f}",
            "| Correct:",
            result["sorted_correctly"]
        )

print("\n=== BIG-O ANALYSIS ===")
print("Bubble Sort:")
print("  Best case: O(n) when optimized and already sorted")
print("  Average case: O(n^2)")
print("  Worst case: O(n^2)")

print("\nSelection Sort:")
print("  Best case: O(n^2)")
print("  Average case: O(n^2)")
print("  Worst case: O(n^2)")

print("\nInsertion Sort:")
print("  Best case: O(n)")
print("  Average case: O(n^2)")
print("  Worst case: O(n^2)")

print("\nSpace complexity for all three algorithms: O(1)")

```
