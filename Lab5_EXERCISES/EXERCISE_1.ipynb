```python
LAST_NAME = "Mabanglo"
STUDENT_ID = "TUPM-26-1847"
FAVORITE_ARTIST = "Zara Larsson"

SEED_DIGIT = int(STUDENT_ID[-1])

ID_SUM = sum(
    int(character)
    for character in STUDENT_ID
    if character.isdigit()
)

NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

ARTIST_SUM = sum(
    ord(character.upper()) - ord("A") + 1
    for character in FAVORITE_ARTIST
    if character.isalpha()
)

print("=== STUDENT-SPECIFIC INPUTS ===")
print("LAST NAME:", LAST_NAME)
print("STUDENT ID:", STUDENT_ID)
print("FAVORITE ARTIST:", FAVORITE_ARTIST)
print("SEED DIGIT:", SEED_DIGIT)
print("ID SUM:", ID_SUM)
print("NAME LENGTH:", NAME_LENGTH)
print("ARTIST LENGTH:", ARTIST_LENGTH)
print("ARTIST SUM:", ARTIST_SUM)
```

```python
record_count = 20

original_records = sorted(
    {
        ID_SUM * 100
        + ARTIST_SUM
        + (index * 17)
        + NAME_LENGTH
        for index in range(record_count)
    }
)

rotation_point = SEED_DIGIT % len(original_records)

rotated_records = (
    original_records[rotation_point:]
    + original_records[:rotation_point]
)

target_index = (
    ID_SUM + ARTIST_SUM
) % len(original_records)

target_incident = original_records[target_index]

print("=== INCIDENT DATA ===")
print("ORIGINAL RECORDS:", original_records)
print("ROTATION POINT:", rotation_point)
print("ROTATED RECORDS:", rotated_records)
print("TARGET INCIDENT:", target_incident)
```

```python
def rotated_binary_search(records, target):
    low = 0
    high = len(records) - 1
    comparisons = 0

    while low <= high:
        middle = (low + high) // 2

        comparisons += 1
        if records[middle] == target:
            return middle, comparisons

        comparisons += 1
        if records[low] <= records[middle]:
            comparisons += 1

            if records[low] <= target < records[middle]:
                high = middle - 1
            else:
                low = middle + 1

        else:
            comparisons += 1

            if records[middle] < target <= records[high]:
                low = middle + 1
            else:
                high = middle - 1

    return -1, comparisons


rotated_position, binary_comparisons = rotated_binary_search(
    rotated_records,
    target_incident
)

print("=== ROTATED BINARY SEARCH ===")

if rotated_position == -1:
    print("TARGET INCIDENT: NOT FOUND")
else:
    print("TARGET INCIDENT:", target_incident)
    print("POSITION:", rotated_position + 1)

print("COMPARISONS:", binary_comparisons)
```

```python
def linear_search(records, target):
    comparisons = 0

    for index, value in enumerate(records):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons


linear_position, linear_comparisons = linear_search(
    rotated_records,
    target_incident
)

print("=== LINEAR SEARCH ===")

if linear_position == -1:
    print("TARGET INCIDENT: NOT FOUND")
else:
    print("TARGET INCIDENT:", target_incident)
    print("POSITION:", linear_position + 1)

print("COMPARISONS:", linear_comparisons)
```

```python
print("=== FINAL EMERGENCY DISPATCH RECORD SEARCH ===")
print("ORIGINAL DATASET:", original_records)
print("ROTATED DATASET:", rotated_records)
print("TARGET INCIDENT:", target_incident)
print("ROTATION POINT:", rotation_point)

print(
    "ROTATED SEARCH POSITION:",
    rotated_position + 1 if rotated_position != -1 else "NOT FOUND"
)

print("ROTATED SEARCH COMPARISONS:", binary_comparisons)
print("LINEAR SEARCH COMPARISONS:", linear_comparisons)

print("\nCOMPLEXITY ANALYSIS")
print("Rotated binary search time complexity: O(log n)")
print("Rotated binary search space complexity: O(1)")
print("Linear search time complexity: O(n)")
print("Linear search space complexity: O(1)")
```
