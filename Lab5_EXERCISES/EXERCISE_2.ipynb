```python
LAST_NAME = "Mabanglo"
STUDENT_ID = "TUPM-26-1847"
FAVORITE_ARTIST = "Zara Larsson"

SEED_DIGIT = int(STUDENT_ID[-1])

ID_SUM = sum(
    int(character)
    for character in STUDENT_ID
    if character.isdigit()
)

NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

ARTIST_SUM = sum(
    ord(character.upper()) - ord("A") + 1
    for character in FAVORITE_ARTIST
    if character.isalpha()
)

# Generate Two Sorted Enrollment Datasets

department_a = sorted(
    SEED_DIGIT * 100
    + ID_SUM
    + index * 13
    for index in range(8)
)

department_b = sorted(
    ARTIST_LENGTH * 100
    + NAME_LENGTH
    + index * 11
    for index in range(7)
)

print("=== ENROLLMENT DATASETS ===")
print("DEPARTMENT A:", department_a)
print("DEPARTMENT B:", department_b)
```

```python

def merge_sorted_records(dataset_a, dataset_b):
    merged = []
    index_a = 0
    index_b = 0
    comparisons = 0

    while index_a < len(dataset_a) and index_b < len(dataset_b):
        comparisons += 1

        if dataset_a[index_a] <= dataset_b[index_b]:
            merged.append(dataset_a[index_a])
            index_a += 1
        else:
            merged.append(dataset_b[index_b])
            index_b += 1

    while index_a < len(dataset_a):
        merged.append(dataset_a[index_a])
        index_a += 1

    while index_b < len(dataset_b):
        merged.append(dataset_b[index_b])
        index_b += 1

    return merged, comparisons


consolidated_records, merge_comparisons = merge_sorted_records(
    department_a,
    department_b
)

print("=== CONSOLIDATED ENROLLMENT RECORDS ===")
print(consolidated_records)
print("COMPARISONS:", merge_comparisons)
```

```python
expected_record_count = len(department_a) + len(department_b)

is_ordered = all(
    consolidated_records[index]
    <= consolidated_records[index + 1]
    for index in range(len(consolidated_records) - 1)
)

correct_record_count = (
    len(consolidated_records) == expected_record_count
)

print("=== VERIFICATION ===")
print("RECORDS BEFORE MERGING:", expected_record_count)
print("RECORDS AFTER MERGING:", len(consolidated_records))
print("ORDER VERIFIED:", is_ordered)
print("RECORD COUNT VERIFIED:", correct_record_count)
```

```python
print("=== FINAL UNIVERSITY ENROLLMENT REPORT ===")
print("DEPARTMENT A DATASET:", department_a)
print("DEPARTMENT B DATASET:", department_b)
print("CONSOLIDATED DATASET:", consolidated_records)
print(
    "NUMBER OF RECORDS BEFORE CONSOLIDATION:",
    expected_record_count
)
print(
    "NUMBER OF RECORDS AFTER CONSOLIDATION:",
    len(consolidated_records)
)
print("COMPARISON COUNT:", merge_comparisons)
print("ORDERING VERIFIED:", is_ordered)
print("RECORD COUNT VERIFIED:", correct_record_count)

print("\nCOMPLEXITY ANALYSIS")
print("Time complexity: O(n + m)")
print("Space complexity: O(n + m)")
```
